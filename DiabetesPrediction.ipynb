```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

diabetes_dataset= pd.read_csv('diabetes_prediction_dataset.csv')
diabetes_dataset.head()
```

```python
Separating the labels X and y
```

```python
X = diabetes_dataset.drop(['diabetes'], axis=1)
y = diabetes_dataset['diabetes']

print(X)
```

```python

print(y)
```

```python
X['smoking_history'].value_counts()
```

```python
Label Encoding for Gender and Smoking_history
0 --> Female 1-->  Male
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
X['gender'] = le.fit_transform(X['gender'])
X['smoking_history'] = X['smoking_history'].replace({'not current':'former'})

X['smoking_history'] = le.fit_transform(X['smoking_history'])

print(X)
```

```python
Data Standarization
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler.fit_transform(X)



```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state = 1)
print(X_train)
print(X_test)
```

```python
Training the Model
```

```python
from xgboost import XGBClassifier
xgb_model = XGBClassifier(ojective='binary:logistic', n_estimators=100, random_state=42)
xgb_model.fit(X_train, y_train)
```

```python
Testing the Model Accuracy Score
```

```python
from sklearn.metrics import accuracy_score
y_pred = xgb_model.predict(X_test)
score = accuracy_score(y_test, y_pred)
score
```

Model Prediction

```python

input_data = (0, 80.0 ,0, 1 ,3 ,25.19 ,6.6 ,140)
input_data_as_numpy_array = np.asarray(input_data)
input_data_reshaped = input_data_as_numpy_array.reshape(1, -1)
std_data = scaler.transform(input_data_reshaped)


prediction = xgb_model.predict(std_data)

print(prediction)

if(prediction == 0):
    print('The person is not Diabetic')
else:
    print('The person is Diabetic')

```
